program to convert the teperature from celsius to fahrenhiets

```python
#Accepting temperature in celsius from the user
c=float(input("Enter the temperature in Celsius"))
```

```python
#convert Celsius into fahrenheit
f=(c*9/5)+32
```

```python
#Displaying temperature in both units
print("Celsius=",c)
print("Fahrenhiets=",f)
```
