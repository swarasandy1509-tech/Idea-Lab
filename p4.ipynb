Program to check whether a number is Positive, negative or zero and even or odd

```python
#Accepts a number from the user
n = int(input("Enter a number: "))
```

```python
#Check whether the number is positive, negative or zero
if n > 0:
    print("Positive")
elif n < 0:
    print("Negative")
else:
    print("Zero")
```

```python
#Check whether the number is even or odd
if n != 0:
    if n % 2 == 0:
        print("Even")
    else:
        print("Odd")
```
