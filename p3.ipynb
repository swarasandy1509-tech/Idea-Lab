Program to accept marks of five subjects and display total marks, Percentage and Result


```python
#Accept mark of five subject form the user
m1 = float(input("Enter marks of subject 1: "))
m2 = float(input("Enter marks of subject 2: "))
m3 = float(input("Enter marks of subject 3: "))
m4 = float(input("Enter marks of subject 4: "))
m5 = float(input("Enter marks of subject 5: "))
```

```python
total = m1 + m2 + m3 + m4 + m5  #Calculate total marks
percentage = total / 5  #Calculate percentage
```

```python
print("Total Marks =", total) #Displays total marks
print("Percentage =", percentage) #Display percentage
if percentage >= 40:
    print("Result: Pass")
else:
    print("Result: Fail")   #check whether student is pass or fail
```
