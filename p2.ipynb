***Program to accepts three decimal numbers from the user and displays.***

```python
#Accepts three decimal numbers as input
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))
```

```python
#Displays addition and Subtraction of three numbers
print("Addition =", a + b + c)
print("Subtraction =", a - b - c)
```

```python
#Displays Multiplication and Division of three numbers
print("Multiplication =", a * b * c)
if b != 0 and c != 0:
    print("Division =", a / b / c)
else:
    print("Division not possible")
```
