Taking Student information including Name, PRN, Branch, College from User


```python
name = input("Enter Student Name: ")  #Accepts Student name from the user
PRN = input("Enter PRN : ") #Accepts Student PRN from the user
branch = input("Enter Branch: ")  #Accepts Branch from the user
college = input("Enter College Name: ") #Accepts college name from the user
```

```python
#Displays Student information
print("\n--- Student Information ---")
print("Student Name:", name)
print("PRN :", PRN)
print("Branch:", branch)
print("College Name:", college)
```
