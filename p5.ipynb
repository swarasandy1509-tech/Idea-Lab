Program to accept marks of five subjects and display total marks, Percentage, Result and *grade*

```python
m1 = float(input("Enter marks of subject 1: "))
m2 = float(input("Enter marks of subject 2: "))
m3 = float(input("Enter marks of subject 3: "))
m4 = float(input("Enter marks of subject 4: "))
m5 = float(input("Enter marks of subject 5: "))
```

```python
total = m1 + m2 + m3 + m4 + m5
percentage = total / 5
print("Total Marks =", total)
print("Percentage =", percentage)
```

```python

if percentage >= 40:
    print("Result: Pass")
else:
    print("Result: Fail")
if percentage >= 75:
    print("Grade: A")
elif percentage >= 60:
    print("Grade: B")
elif percentage >= 50:
    print("Grade: C")
elif percentage >= 40:
    print("Grade: D")
else:
    print("Grade: E")
```
